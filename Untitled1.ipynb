```python
import pandas as pd
import matplotlib.pyplot as plt

# Correct path to the file
df = pd.read_csv('/content/Air_Quality.csv')

# Filtering using actual values ('city' instead of 'City Center')
city_center_data = df[df['Location'] == 'city']

# Grouping by exact column names from the CSV
avg_pollution = df.groupby('Location')[['Air _Quality_Category', 'PM2.5_Level']].mean(numeric_only=True)

plt.figure(figsize=(8, 4))
# Plotting using correct column name 'Air _Quality_Category'
plt.plot(city_center_data['Date'], city_center_data['Air _Quality_Category'], marker='o', color='red')
plt.title('Chart 1: Air _Quality_Category Trend at City')
plt.xlabel('Date')
plt.ylabel('Air Quality Category')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

# Corrected the column name capitalization to 'PM2.5_Level'
avg_pollution['PM2.5_Level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure data is loaded
df = pd.read_csv('/content/Air_Quality.csv')

plt.figure(figsize=(8, 4))

# Corrected the column name capitalization to 'PM2.5_Level'
plt.scatter(df['Date'], df['PM2.5_Level'], color='purple', s=80)

plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset locally to ensure 'df' is defined
df = pd.read_csv('/content/Air_Quality.csv')

# chart 4: Histogram (Distribution of all PM2.5 readings)
plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_Level'], bins=20, color='green', edgecolor='black')
plt.title('Chart 4: Distribution of PM2.5 Readings')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
#chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='barh', color='teal')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Max PM2.5')
plt.ylabel('Location')
plt.show()
```

```python
#chart 6: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='teal')
plt.title('Chart 6: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Peak PM2.5 Level')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
